```python
# ============================================================ 
# INICIO DEL SCRIPT 
# ============================================================ 
# Importa el módulo pyplot de Matplotlib. 
# Usamos el alias plt para escribir instrucciones más cortas. 
import matplotlib.pyplot as plt 
# Crea una lista llamada valores. 
# La lista contiene cinco números que serán analizados en el histograma. 
valores = [10, 15, 12, 18, 20] # Crea el histograma. 
# valores: contiene los datos que se van a representar. 
# bins=5: divide el rango de los datos en cinco intervalos. 
# edgecolor="black": dibuja el borde de cada barra en color negro. 
plt.hist(valores, bins=5, edgecolor="black") 
# Coloca el texto "Valores" como nombre del eje horizontal X. 
plt.xlabel("Valores") 
# Coloca el texto "Frecuencia" como nombre del eje vertical Y. 
plt.ylabel("Frecuencia") 
# Agrega un título en la parte superior del gráfico.
plt.title("Histograma de 5 valores") 
# Muestra el gráfico en pantalla. 
plt.show() 
# ============================================================ 
# FIN DEL SCRIPT 
# ============================================================
```
